# NYC Taxi Dataset

Viajes en taxi en la ciudad de Nueva York (sólo taxis amarillos) en los años 2015 y 2016.

**Taxis amarillos**: estos son los famosos taxis amarillos de la ciudad de Nueva York que prestan servicio exclusivamente mediante señas en la calle. El número de taxis está limitado por la cantidad finita de licencias emitidas por la TLC. Se accede a este modo de transporte colocándose en la calle y llamando con la mano a un taxi disponible. Las recogidas no están prearregladas.

**Vehículos de alquiler (FHV)**: el servicio FHV se accede mediante una reserva previa con un despachador o una compañía de limusinas. Estos FHV no están autorizados a recoger pasajeros mediante señas en la calle, ya que esos viajes no se consideran pre-reservados.

**Taxis verdes**: Street Hail Livery (SHL). El programa SHL permitirá a los propietarios de vehículos de alquiler (VTC) licenciar y equipar sus vehículos con la identidad de taxi verde del distrito, instalar taxímetros y máquinas para tarjetas de crédito y, en última instancia, obtener el derecho a aceptar señas en la calle además de viajes pre-reservados.

En este dataset se consideran únicamente los datos de taxis amarillos de los meses de enero de 2015 y de enero a marzo de 2016.

```python
from pyspark.sql import DataFrame as DF
from pyspark.sql import functions as F


taxi_csv_2016_01_path = "abfss://datos@matermcb001sta.dfs.core.windows.net/yellow_tripdata_2016-01.csv"

taxi_csv_2016_02_path = "abfss://datos@matermcb001sta.dfs.core.windows.net/yellow_tripdata_2016-02.csv"

taxi_csv_2016_03_path = "abfss://datos@matermcb001sta.dfs.core.windows.net/yellow_tripdata_2016-03.csv"

def read_and_clean_csv(path: str) -> DF:

    transformed_df = (
        spark.read
        .format("csv")
        .option("header", "true")
        .option("inferSchema", "true")
        .option("adls.account.name", "matermcb001sta")
        .option(
            "fs.azure.account.key.matermcb001sta.dfs.core.windows.net",
            "a+X6TcqNgcaSv+lTwr0jsVKuiFCRQ0UL5KfOcXV5b80TwDol3xW1VcLzvD7tkb+ZE5kyMxo8zxuz+AStaCTWOQ=="
        )

        .load(path)

        # duración del viaje en segundos
        .withColumn(
            "trip_duration_s",
            F.timestamp_diff(
                "SECOND",
                F.col("tpep_pickup_datetime"),
                F.col("tpep_dropoff_datetime")
            )
        )

        # filtros
        .filter(F.col("trip_distance") > 0)
        .filter(F.col("total_amount") > 0)
        .filter(F.col("fare_amount") > 0)
        .filter(F.col("total_amount") >= F.col("fare_amount"))
        .filter(F.col("trip_distance") >= 0.62)
        .filter(F.col("trip_duration_s") >= 120)

        # identificador único
        .withColumn(
            "trip_id",
            F.abs(
                F.xxhash64(
                    F.concat(
                        "tpep_pickup_datetime",
                        "tpep_dropoff_datetime"
                    )
                )
            )
        )

        # horas
        .withColumn(
            "pickup_hour",
            F.hour("tpep_pickup_datetime")
        )

        .withColumn(
            "dropoff_hour",
            F.hour("tpep_dropoff_datetime")
        )

        # día de la semana numérico
        .withColumn(
            "day_of_week",
            F.dayofweek("tpep_pickup_datetime")
        )

        # traducir día semana
        .withColumn(
            "day_of_week",
            F.when(F.col("day_of_week") == 1, "domingo")
            .when(F.col("day_of_week") == 2, "lunes")
            .when(F.col("day_of_week") == 3, "martes")
            .when(F.col("day_of_week") == 4, "miercoles")
            .when(F.col("day_of_week") == 5, "jueves")
            .when(F.col("day_of_week") == 6, "viernes")
            .when(F.col("day_of_week") == 7, "sabado")
        )

        # velocidad media
        .withColumn(
            "speed",
            F.round(
                (F.col("trip_distance") / F.col("trip_duration_s")) * 3600,
                2
            )
        )

        # porcentaje propina
        .withColumn(
            "tip_perc",
            F.col("tip_amount") / F.col("total_amount")
        )
    )

    return transformed_df


nyc_trips_df = (
    read_and_clean_csv(taxi_csv_2016_01_path)
    .union(read_and_clean_csv(taxi_csv_2016_02_path))
    .union(read_and_clean_csv(taxi_csv_2016_03_path))
.cache()
)

```

        * La función `F.coalesce(col1, col2, ...)` devuelve un objeto columna en el cual, en cada fila, estará el valor de la primera columna que no sea NULL que encuentre cuando va recorriendo en esa fila las columnas que le hemos pasado como argumentos (las irá recorriendo en el orden en el que se las hemos pasado).

A veces incluso indica un rango de días en los que se repetirá el evento, pero en esos casos basta con extraer la primera fecha. Se pide:

4. Extraer el mes y el día del evento de la columna `"Event Description"` mediante la expresión regular indicada en la siguiente celda. Se debe usar con `F.regexp_extract`
para crear nuevas columnas `"event_month"`, `"event_day"`, y `"event_year"`, las tres de tipo string, pasando en cada caso a la función `regexp_extract` el argumento `idx` con valores 1, 2 y 3 respectivamente, para extraer el contenido del grupo de matching (paréntesis) primero, segundo y tercero de la expresión regular, que son el mes, el día del mes, y el año. Por último, crear una nueva columna `"event_date"` de tipo Date, concatenando las tres anteriores por ese orden, con un espacio en blanco, usando `F.try_to_date(F.concat_ws(" ", ...), "MMMM d yyyy")`. De esta forma está asegurado que la regex solamente captura la primera fecha, en caso de que haya un rango de fechas. Las columnas `event_month`, `event_day` y `event_year` deben eliminarse después de crear la columna `event_date`. Si no se encuentra la fecha en `Event Description`, la columna `event_date` habrá quedado a NULL, pero en ese caso debe reemplazarse por el valor de la fecha indicada en `Close Time` tras convertirlo a Date con `F.try_to_date`. Puedes usar `F.coalesce` para esto, poniendo en primer lugar la columna `event_date` y en segundo lugar la conversión a Date de `Close Time`.
    * **NOTA**: esta función se introdujo nueva en Spark 4.1.0 y por tanto, requiere que el Databricks Runtime seleccionado para el cluster sea el 18.0 que trae Spark 4.1.0.

La lectura y todas estas transformaciones (todas son withColumn) deben estar encadenadas, sin usar variables intermedias, y el resultado debe guardarse en la variable `raw_events_df`, sin cachear.
Para el último paso de crear event_date, puedes encadenar una transformación más a la que ya tenías, sin necesidad de crear una nueva variable intermedia, para reemplazar los posibles nulos de event_date por el resultado de convertir Close Time a Date.

```python
from pyspark.sql import SparkSession, functions as F

regex_event_date = r'(?i)\b(January|February|March|April|May|June|July|August|September|October|November|December)\s+(\d{1}|\d{2})(?:st|nd|rd|th)?\s*,\s*(\d{4})\b'


csv_path = "abfss://datos@matermcb001sta.dfs.core.windows.net/511_NY_Events_raw.csv"

raw_events_df = (
    spark.read
    .format("csv")
    .option("header", "true")
    .option("inferSchema", "true")
    .option("sep", ";")
    .option("adls.account.name", "matermcb001sta")
    .option(
            "fs.azure.account.key.matermcb001sta.dfs.core.windows.net",
            "a+X6TcqNgcaSv+lTwr0jsVKuiFCRQ0UL5KfOcXV5b80TwDol3xW1VcLzvD7tkb+ZE5kyMxo8zxuz+AStaCTWOQ=="
        )
    .load(csv_path)

    # Close Time
    .withColumn(
        "Close Time",
        F.coalesce(
            F.try_to_timestamp(
                F.col("Close Time"),
                F.lit("MM/dd/yyyy hh:mm:ss a")
            ),
            F.to_timestamp(
                F.col("Close Time"),
                "MM/dd/yyyy H:mm"
            )
        )
    )

    # Create Time
    .withColumn(
        "Create Time",
        F.coalesce(
            F.try_to_timestamp(
                F.col("Create Time"),
                F.lit("MM/dd/yyyy hh:mm:ss a")
            ),
            F.to_timestamp(
                F.col("Create Time"),
                "MM/dd/yyyy H:mm"
            )
        )
    )

    # Latitude
    .withColumn(
        "Latitude",
        F.regexp_replace(
            F.col("Latitude"),
            ",",
            "."
        ).cast("double")
    )

    # Longitude
    .withColumn(
        "Longitude",
        F.regexp_replace(
            F.col("Longitude"),
            ",",
            "."
        ).cast("double")
    )

    # event_month
    .withColumn(
        "event_month",
        F.regexp_extract(
            F.col("Event Description"),
            regex_event_date,
            1
        )
    )

    # event_day
    .withColumn(
        "event_day",
        F.regexp_extract(
            F.col("Event Description"),
            regex_event_date,
            2
        )
    )

    # event_year
    .withColumn(
        "event_year",
        F.regexp_extract(
            F.col("Event Description"),
            regex_event_date,
            3
        )
    )

    # event_date
    .withColumn(
        "event_date",
        F.coalesce(
            F.try_to_date(
                F.concat_ws(
                    " ",
                    F.col("event_month"),
                    F.col("event_day"),
                    F.col("event_year")
                ),
                "MMMM d yyyy"
            ),
            F.try_to_date(F.col("Close Time"))
        )
    )

    # borrar auxiliares
    .drop(
        "event_month",
        "event_day",
        "event_year"
    )
)
```

```python
assert(raw_events_df.count() == 13715)

# Casuística 1: event_date se ha extraído correctamente de Event Description
r1 = raw_events_df.where(F.col("Event Description").contains("Tuesday June 2nd"))\
                  .select(F.date_format("event_date", "yyyy-MM-dd").alias("event_date")).first()
assert(r1.event_date == "2020-06-02")

# Casuística 2: event_date no se ha podido extraer de Event Description, por lo que se ha deducido de Close Time
r2 = raw_events_df.where( (F.date_format("Close Time", "yyyy-MM-dd") == "2022-05-28") &
                          (F.col("Facility Name") == "Meadowlands Racetrack"))\
                  .select(F.date_format("event_date", "yyyy-MM-dd").alias("event_date")).first()
assert(r2.event_date == "2022-05-28")
```

Tras esto, en los casos en los que `event_end_ts` fuese nulo (no se pudo extraer la hora de fin con la regex, la cual devolvió cadena vacía, y por eso su conversión a timestamp resulta en nulo), establecer `event_end_ts` como el valor de `Close Time` (convertido a timestamp en un apartado previo). Esto puede hacerse usando `F.coalesce` (consulta la documentación de la función). En los casos en los `event_start_ts` fuese nulo (no se pudo extraer la hora de inicio con la regex, devolvió cadena vacía y por eso su conversión a timestamp resulta en nulo), debemos crear `event_start_ts` como el resultado de restar 2 horas a `"event_end_ts"`, incluso para los casos donde `event_end_ts` fue copiado de `Close TIme`, utilizando para ello `F.timestamp_add("HOUR", -2, "event_start_ts")`. Estamos asumiendo que los eventos duran dos horas para deducir la hora de inicio. Aquí puede usarse también `F.coalesce`.

4. Eliminar aquellas filas en las que `event_date`, `event_start_ts` o `event_end_ts` sean nulos cualquiera de los tres.
5. Eliminar filas duplicadas considerando únicamente las columnas `Event Description` y `event_date` simultáneamente.
6. El resultado de todas estas transformaciones (encadenadas) debe guardarse en un nuevo DataFrame llamado `events_df`, que debe quedar cacheado. No debe utilizarse ninguna variable intermedia, ni asignaciones intermedias del estilo `event_df = event_df.funcion(...)` (serán penalizadas).


```python
regex_event_time_both = r'(?i)(?:from\s+)?(\d{1,2}:\d{2}\s*(?:AM|PM|A\.M\.|P\.M\.))\s*(?:thru|until|to|-)\s*(\d{1,2}:\d{2}\s*(?:AM|PM|A\.M\.|P\.M\.))'
regex_event_time_until = r'(?i)\s*until\s*(\d{1,2}:\d{2}\s*(?:AM|PM|A\.M\.|P\.M\.))'

events_df = (
    raw_events_df

    # Extracted Start Time
    .withColumn(
        "Extracted Start Time",
        F.regexp_extract(
            F.col("Event Description"),
            regex_event_time_both,
            1
        )
    )

    # Extracted End Time
    .withColumn(
        "Extracted End Time",
        F.regexp_extract(
            F.col("Event Description"),
            regex_event_time_both,
            2
        )
    )

    # casos "until"
    .withColumn(
        "Extracted End Time",
        F.when(
            F.col("Extracted End Time") == "",
            F.regexp_extract(
                F.col("Event Description"),
                regex_event_time_until,
                1
            )
        ).otherwise(
            F.col("Extracted End Time")
        )
    )

    # quitar puntos
    .withColumn(
        "Extracted Start Time",
        F.regexp_replace(
            F.col("Extracted Start Time"),
            r"\.",
            ""
        )
    )

    .withColumn(
        "Extracted End Time",
        F.regexp_replace(
            F.col("Extracted End Time"),
            r"\.",
            ""
        )
    )

    # event_end_ts
    .withColumn(
        "event_end_ts",
        F.try_to_timestamp(
            F.concat_ws(
                " ",
                F.col("event_date"),
                F.col("Extracted End Time")
            ),
            F.lit("yyyy-MM-dd h:mm a")
        )
    )

    # event_start_ts
    .withColumn(
        "event_start_ts",
        F.try_to_timestamp(
            F.concat_ws(
                " ",
                F.col("event_date"),
                F.col("Extracted Start Time")
            ),
            F.lit("yyyy-MM-dd h:mm a")
        )
    )

    # si termina al día siguiente
    .withColumn(
        "event_end_ts",
        F.when(
            F.col("event_end_ts") < F.col("event_start_ts"),
            F.date_add(
                F.col("event_end_ts"),
                1
            )
        ).otherwise(
            F.col("event_end_ts")
        )
    )

    # rellenar event_end_ts
    .withColumn(
        "event_end_ts",
        F.coalesce(
            F.col("event_end_ts"),
            F.col("Close Time")
        )
    )

    # rellenar event_start_ts
    .withColumn(
        "event_start_ts",
        F.coalesce(
            F.col("event_start_ts"),
            F.expr("event_end_ts - INTERVAL 2 HOURS"
        )
        )
    )

    # eliminar nulos
    .filter(
        F.col("event_date").isNotNull()
    )

    .filter(
        F.col("event_start_ts").isNotNull()
    )

    .filter(
        F.col("event_end_ts").isNotNull()
    )

    # eliminar duplicados
    .dropDuplicates(
        ["Event Description", "event_date"]
    )

    .cache()
)
```

Para esto, primero debes crear una función `manhattan_distance_meters` que calcule la distancia de Manhattan en metros entre dos puntos geográficos dados por sus latitudes y longitudes, y devuelva un objeto Column con esa distancia.
  * Esta función debe recibir solamente 4 argumentos, que serán 4 strings con los nombres de las columnas latitud, longitud de un punto y latitud, longitud del otro punto, y debe operar todo el rato con objetos Column que no están ligados a ningún DataFrame (por eso no recibe ningún DF como argumento).
  * Para el cálculo de la distancia de Manhattan, puedes calcular por separado la componente Norte-Sur (diferencia de latitud) y la componente Este-Oeste como sigue, y devolver el objeto Column con la suma de ambas.
    * Diferencia norte-sur en metros: valor absoluto de la diferencia de latitudes (tras expresar primero ambas latitudes en radianes con `F.radians(colname)`), multiplicada por el radio de la Tierra en metros (6371000 m).     * Diferencia este-oeste en metros: valor absoluto de la diferencia de longitudes (tras expresar primero ambas longitudes en radianes con `F.radians(colname)`), multiplicada por el **coseno de la media de las dos latitudes (en radianes)** de ambos puntos y multiplicado por el radio de la Tierra en metros. Es decir, abs(lon2_rad - lon1_rad) * cos((lat1_rad + lat2_rad) / 2) * 6371000.
    * Recuerda que los objetos Column en Spark pueden operarse con sumas, restas, multiplicaciones, y que existe `F.abs`, `F.radians` y `F.cos` que reciben columnas como argumento y devuelven columnas.
* Una vez creada la función, filtrar `nyc_trips_filtered` para quedarnos solamente con los viajes que tengan el punto de llegada a una distancia de Manhattan menor o igual a 100 metros de la ubicación del evento, haciendo un join entre `nyc_trips_filtered` y el resultado de quitar duplicados por longitud y latitud a `events_df_filtered`, usando como condición de join una llamada a la distancia de Manhattan entre el punto de llegada del viaje y la ubicación del evento, que debe ser menor o igual a 100 metros.
  * OJO: estamos usando join para "filtrar" porque no nos interesan las columnas de la derecha... ¿qué tipo de join necesitas para no añadir columnas? El resultado se guardará en `nyc_trips_cercanos_sedes`, el cual debe quedar cacheado. Estos viajes son solamente los que terminan cerca de las ubicaciones de algún evento, pero sin tener en cuenta la hora del evento, por lo que todavía no sabemos si realmente han ido a esos eventos o simplemente han coincidido en la ubicación pero no en el horario. Será útil para comparar posteriormente con los viajes que sí coinciden también en el horario del evento, para ver si realmente hay un aumento de la demanda en las inmediaciones de los eventos durante su horario.
* Finalmente, partiendo de `nyc_trips_cercanos_sedes`, quedarnos solamente con los viajes que llegan a las inmediaciones de un evento (los del paso anterior) y que además llegan durante el horario del evento, es decir, que la diferencia en minutos entre la hora de llegada del viaje y la hora de inicio del evento esté entre 0 y 20 minutos. Para esto, se debe hacer un join entre `nyc_trips_cercanos_sedes` y `events_df_filtered`, usando como condición de join que la diferencia en minutos entre la hora de llegada del viaje (`tpep_dropoff_datetime`) y la hora de inicio del evento (`event_start_ts`) esté entre 0 y 20 minutos. No queremos perder ningún viaje, solo saber si coincide con el horario de algún evento o no, por lo que el tipo de join debe ser el adecuado para esto.
  * PISTA: la condición de join es que `F.timestamp_diff("MINUTE", ..., ...).between(..., ...)`. En el DF resultante deben quitarse los duplicados por `trip_id`, y después deben rellenarse los valores nulos en la columna "Event type" por el valor "No event" (casos donde no se encontró matching al ser un viaje que terminó cerca de la sede pero no en la franja horaria adecuada). El DF resultante se guardará en `nyc_trips_eventos`, el cual debe quedar cacheado. En este DataFrame ya sí podemos asumir que los viajes que aparecen allí han ido a esos eventos, porque han coincidido tanto en la ubicación como en el horario.

```python
def manhattan_distance_meters(lat1_col: str, lon1_col: str, lat2_col: str, lon2_col: str):

    # coordenadas en radianes
    lat1_rad = F.radians(F.col(lat1_col))
    lon1_rad = F.radians(F.col(lon1_col))

    lat2_rad = F.radians(F.col(lat2_col))
    lon2_rad = F.radians(F.col(lon2_col))

    # componente norte-sur
    north_south = (
        F.abs(lat2_rad - lat1_rad) * 6371000
    )

    # componente este-oeste
    east_west = (
        F.abs(lon2_rad - lon1_rad) *
        F.cos((lat1_rad + lat2_rad) / 2) *
        6371000
    )

    # distancia Manhattan
    return north_south + east_west


events_df_filtered = (
    events_df.filter(
        (F.col("event_date") >= "2016-01-01") &
        (F.col("event_date") <= "2016-03-31") &
        (F.hour("event_start_ts") >= 18)
    )
    .cache()
)

nyc_trips_filtered = (
    inlier_locations_df.filter(
        F.hour("tpep_dropoff_datetime") >= 17
    )
    .cache()
)

   

nyc_trips_cercanos_sedes = (
    nyc_trips_filtered.join(
        events_df_filtered.dropDuplicates(
            ["Latitude", "Longitude"]
        ),
        manhattan_distance_meters(
            "dropoff_latitude",
            "dropoff_longitude",
            "Latitude",
            "Longitude"
        ) <= 100,
        "left_semi"
    )
    .cache()
)

nyc_trips_eventos = (
    nyc_trips_cercanos_sedes
    .join(
        events_df_filtered,

        F.timestamp_diff(
            "MINUTE",
            F.col("tpep_dropoff_datetime"),
            F.col("event_start_ts")
        ).between(0, 20),

        "left"
    )
    .orderBy("Event type")

    .dropDuplicates(["trip_id"])

    .fillna({
        "Event type": "No event"
    })

    .cache()
)
```

El resultado de esta agregación se guardará en una columna llamada `avg_tip_perc`. Es decir, `F.round(F.mean(F.when(...., F.col("tip_perc")).otherwise(None)), 2).alias(...)`.
* Tras las agregaciones, ordenar el DF obtenido ascendentemente por la hora de llegada.

El DF resultante debe quedar guardado en la variable `trips_statistics_df`.

Aunque no sea evaluable, analiza las siguientes cuestiones:
* ¿Por qué hay tantas casillas vacías (con valor nulo) en el resultado?
* ¿Qué tipo de eventos tienen más viajes asociados, y cuáles menos?
* ¿Hay alguna hora del día donde haya más viajes asociados a eventos que en otras horas?

```python
trips_statistics_df = (
    nyc_trips_eventos

    .groupBy("dropoff_hour")
    .pivot("Event type")
    .agg(
        # número viajes
        F.count("trip_id").alias("num_trips"),
        # distancia media
        F.round(
            F.avg("trip_distance"),
            2
        ).alias("avg_distance"),
        # duración media
        F.round(
            F.avg("trip_duration_s"),
            2
        ).alias("avg_duration_s"),
        # porcentaje pago tarjeta
        F.round(
            F.avg(
                (
                    F.col("payment_type") == 1
                ).cast("int")
            ),
            2
        ).alias("card_payment_perc"),
        # media tip_perc solo tarjeta
        F.round(
            F.mean(
                F.when(
                    F.col("payment_type") == 1,
                    F.col("tip_perc")
                ).otherwise(None)
            ),
            2
        ).alias("avg_tip_perc")
    )
    .orderBy("dropoff_hour")

    .cache()
)
```

```python
trips_agg_statistics_df = (
    nyc_trips_eventos
    .groupBy("Event type")
    .agg(
        # número viajes
        F.count("trip_id").alias("num_trips"),
        # distancia media
        F.round(
            F.avg("trip_distance"),
            2
        ).alias("avg_distance"),
        # distancia mediana
        F.round(
            F.median("trip_distance"),
            2
        ).alias("med_distance"),
        # duración media
        F.round(
            F.avg("trip_duration_s"),
            2
        ).alias("avg_duration_s"),
        # duración mediana
        F.round(
            F.median("trip_duration_s"),
            2
        ).alias("med_duration_s"),
        # porcentaje pago tarjeta
        F.round(
            F.avg(
                (
                    F.col("payment_type") == 1
                ).cast("int")
            ),
            2
        ).alias("card_payment_perc"),
        # promedio propina solo tarjeta
        F.round(
            F.mean(
                F.when(
                    F.col("payment_type") == 1,
                    F.col("tip_perc")
                ).otherwise(None)
            ),
            2
        ).alias("avg_tip_perc")
    )
    .cache()
)
```

```python
trips_agg_statistics_pd = trips_agg_statistics_df.toPandas()
assert(trips_agg_statistics_pd.loc[trips_agg_statistics_pd["Event type"] == "No event", "num_trips"].values[0] == 166428)
assert(trips_agg_statistics_pd.loc[trips_agg_statistics_pd["Event type"] == "concert", "avg_distance"].values[0] in [1.87, 1.88])
assert(trips_agg_statistics_pd.loc[trips_agg_statistics_pd["Event type"] == "boxing", "card_payment_perc"].values[0] == 0.38)
```
